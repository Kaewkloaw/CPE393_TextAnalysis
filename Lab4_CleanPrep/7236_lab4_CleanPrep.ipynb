```python
import re
import requests
import string
from bs4 import BeautifulSoup
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

ดึงข้อความจาก url

```python
url = 'https://stackoverflow.com/questions/38488442/razor-framework-backend-or-frontend'
page = requests.get(url)
soup = BeautifulSoup(page.text, 'html')
```

 Web Scraping

```python
import requests

# Question ID
question_id = 38488442

# --- Fetch the question ---
question_url = f'https://api.stackexchange.com/2.3/questions/{question_id}'
question_params = {
    'order': 'desc',
    'sort': 'activity',
    'site': 'stackoverflow',
    'filter': 'withbody'
}

q_response = requests.get(question_url, params=question_params)
q_data = q_response.json()

if 'items' in q_data and q_data['items']:
    question = q_data['items'][0]
    print("=== Question ===")
    print("Title:", question['title'])
    print("Body:", question['body'])
    print("Score:", question['score'])
    print("Asked by:", question['owner'].get('display_name', 'Unknown'))
    print("="*50, "\n")
else:
    print("Question not found")

# --- Fetch the answers ---
answer_url = f'https://api.stackexchange.com/2.3/questions/{question_id}/answers'
answer_params = {
    'order': 'desc',
    'sort': 'votes',
    'site': 'stackoverflow',
    'filter': 'withbody'
}

a_response = requests.get(answer_url, params=answer_params)
a_data = a_response.json()

answers = a_data.get('items', [])
if answers:
    for i, ans in enumerate(answers, start=1):
        print(f"--- Answer {i} ---")
        print("Body:", ans['body'])
        print("Score:", ans['score'])
        print("Answered by:", ans['owner'].get('display_name', 'Unknown'))
        print("-"*50, "\n")
else:
    print("No answers found")
```

Question *Part*

```python
div_tags = soup.find_all('div', class_='s-prose js-post-body', attrs={'itemprop': 'text'})
question = div_tags[0].text.strip('\n')
print(question)
```

1.cleaning

```python
#clean_punctuation
def clean_punctuation(text):
    punctuation_pattern = re.compile(r'[^\w\s]|_')
    cleaned_text = re.sub(punctuation_pattern, '', text)

    return cleaned_text

question = clean_punctuation(question)
question = question.replace('\n', ' ')
question
```

```python
# lower character
question = question.lower()
print(question)
```

```python
# remove stopwords
words = word_tokenize(question)

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.lower() not in stop_words]
question_cleaning_text = " ".join(filtered_words)
print(question_cleaning_text)
```

```python
# remove special characters
question_cleaning_text = re.sub(r'[^a-zA-Z0-9\s]', '', question_cleaning_text)
question_cleaning_text = re.sub(r'\s+', ' ', question_cleaning_text).strip()

print(question_cleaning_text)
```

```python
# Data masking
def masking(text):
    prog_pattern = re.compile(r'Razor|Aspnet')
    masked_text = re.sub(prog_pattern, '[FRAMEWORK]', text)

    return masked_text
question_cleaning_text = masking(question_cleaning_text)
print(question_cleaning_text)
```

2. Pre-Processing

```python
# Lemmatizer
lemmatizer = WordNetLemmatizer()
words = question_cleaning_text.split(' ')
question_preprocessing_text = ' '.join([lemmatizer.lemmatize(word, pos='n') for word in words])
question_preprocessing_text = ' '.join([lemmatizer.lemmatize(word, pos='v') for word in words])
print(question_preprocessing_text)
```

```python
# POS-tag
from nltk import pos_tag
nltk.download('averaged_perceptron_tagger')
tokens_for_pos = nltk.word_tokenize(question_preprocessing_text)
pos_tags = pos_tag(tokens_for_pos)

print(pos_tags)
```

Answer *Part*

```python
ans_zone = soup.find_all('div', class_='answercell post-layout--right')
answers = [elem.find_all('p') for elem in ans_zone]
answers
```

```python
result = {}

# ใช้ for loop แบ่งข้อมูลและใส่ id
for i in range(len(answers)):
    key = f"id_{i+1}"
    result[key] = answers[i]

# แสดงผล
import pprint
pprint.pprint(result)
```

1.cleaning

```python
# remove html tags
def clean_html_tags(text):
    clean_text = text.get_text(separator=' ')
    clean_text = re.sub(r'\s+' , ' ' , clean_text).strip()

    return clean_text
```

```python
def masking(text):
    return re.sub(r'razor', '[framwork]', text)

stop_words = set(stopwords.words('english'))
preprocessed_result = {}

for key, contents in result.items():
    preprocessed_result[key] = []
    for text in contents:
        # remove html tags
        cleaned_text = clean_html_tags(text)
        # lowercase
        question = cleaned_text.lower()
        # tokenize
        words = word_tokenize(question)
        # remove stopwords
        filtered_words = [word for word in words if word not in stop_words]
        # join words
        question_cleaning_text = " ".join(filtered_words)
        # remove special characters
        question_cleaning_text = re.sub(r'[^a-zA-Z0-9\s]', '', question_cleaning_text)
        question_cleaning_text = re.sub(r'\s+', ' ', question_cleaning_text).strip()
        # data masking
        question_cleaning_text = masking(question_cleaning_text)
        # append to result
        preprocessed_result[key].append(question_cleaning_text)

pprint.pprint(preprocessed_result)
```

2. Pre-Processing

```python
for key, contents in preprocessed_result.items():
    print(f"\n--- Processing {key} ---")
    for i, text in enumerate(contents):
        # Lemmatization สำหรับ noun
        words = text.split(' ')
        lemmatized_n = [lemmatizer.lemmatize(word, pos='n') for word in words]
        # Lemmatization สำหรับ verb
        lemmatized_nv = [lemmatizer.lemmatize(word, pos='v') for word in lemmatized_n]
        question_preprocessing_text = ' '.join(lemmatized_nv)
        print(f"\nOriginal Text: {text}")
        print(f"Lemmatized Text: {question_preprocessing_text}")

        # POS tagging
        tokens_for_pos = word_tokenize(question_preprocessing_text)
        pos_tags = pos_tag(tokens_for_pos)
        print(f"POS Tags: {pos_tags}")
```
